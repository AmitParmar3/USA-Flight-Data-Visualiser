```python
# Convert BTS HHMM values into hours and minutes
def time_to_timedelta(series):
    series = series.astype("Int64")
    
    hours = series // 100
    minutes = series % 100
    
    return pd.to_timedelta(hours, unit="h") + pd.to_timedelta(minutes, unit="m")


# Create scheduled timestamps
df["SCHEDULED_DEP_TS"] = (
    df["FL_DATE"] + time_to_timedelta(df["CRS_DEP_TIME"])
)

df["SCHEDULED_ARR_TS"] = (
    df["FL_DATE"] + time_to_timedelta(df["CRS_ARR_TIME"])
)


# Create actual timestamps
df["ACTUAL_DEP_TS"] = (
    df["FL_DATE"] + time_to_timedelta(df["DEP_TIME"])
)

df["ACTUAL_ARR_TS"] = (
    df["FL_DATE"] + time_to_timedelta(df["ARR_TIME"])
)


# Correct overnight scheduled arrivals
overnight_scheduled = (
    df["SCHEDULED_ARR_TS"] < df["SCHEDULED_DEP_TS"]
)

df.loc[overnight_scheduled, "SCHEDULED_ARR_TS"] += pd.Timedelta(days=1)


# Correct overnight actual arrivals
overnight_actual = (
    df["ACTUAL_ARR_TS"] < df["ACTUAL_DEP_TS"]
)

df.loc[overnight_actual, "ACTUAL_ARR_TS"] += pd.Timedelta(days=1)
```

```python
df["CALCULATED_SCHEDULED_DURATION_MIN"] = (
    df["SCHEDULED_ARR_TS"] - df["SCHEDULED_DEP_TS"]
).dt.total_seconds() / 60

df["CALCULATED_ACTUAL_DURATION_MIN"] = (
    df["ACTUAL_ARR_TS"] - df["ACTUAL_DEP_TS"]
).dt.total_seconds() / 60

df[[
    "CRS_ELAPSED_TIME",
    "CALCULATED_SCHEDULED_DURATION_MIN",
    "ACTUAL_ELAPSED_TIME",
    "CALCULATED_ACTUAL_DURATION_MIN"
]].head(20)
```

```python
df["SCHEDULED_DURATION_DIFF"] = (
    df["CALCULATED_SCHEDULED_DURATION_MIN"]
    - df["CRS_ELAPSED_TIME"]
)

df["ACTUAL_DURATION_DIFF"] = (
    df["CALCULATED_ACTUAL_DURATION_MIN"]
    - df["ACTUAL_ELAPSED_TIME"]
)
```
